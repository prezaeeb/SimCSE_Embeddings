# compute TP, FP,TN, FN


```python
def perf_measure(y_actual, y_hat):
    TP = 0
    FP = 0
    TN = 0
    FN = 0

    for i in range(len(y_hat)): 
        if y_actual[i]==y_hat[i]=="Pos":
           TP += 1
        if y_hat[i]=="Pos" and y_actual[i]!=y_hat[i]:
           FP += 1
        if y_actual[i]==y_hat[i]=="neg":
           TN += 1
        if y_hat[i]=="neg" and y_actual[i]!=y_hat[i]:
           FN += 1

    return(TP, FP, TN, FN)
```
